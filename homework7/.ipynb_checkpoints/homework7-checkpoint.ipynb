```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    y = A*np.sin(w*x)
    return y

x = np.linspace(0, 2*np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

fig, ax = plt.subplots(1, 3, figsize=(25, 5))

ax1 = ax[0]
fixed_w = 1
for A in amplitude:
    y = make_sine_wave(x, A, fixed_w)
    ax1.plot(x, y, label=(f"A={A}"))
ax1.set_title("Varying Amplitude")
ax1.set_xlabel("x")
ax1.set_ylabel("y")
ax1.legend()

ax2 = ax[1]
fixed_a = 1
for w in frequency:
    y = make_sine_wave(x, fixed_a, w)
    ax2.plot(x, y, label=(f"w={w}"))
ax2.set_title("Varying Frequency")
ax2.set_xlabel("x")
ax2.set_ylabel("y")
ax2.legend()

ax3 = ax[2]
fixed_w = 1
for A, w in zip(amplitude, frequency):
    y = make_sine_wave(x, A, w)
    ax3.plot(x, y, label=(f"A={A}, w={w}"))
ax3.set_title("Varying Amplitude and Frequency")
ax3.set_xlabel("x")
ax3.set_ylabel("y")
ax3.legend()

plt.show
```
