```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)
```

```python
print(np.shape(data)) #2642 rows

magnitude = data[:,0]
temperature = data[:,1]

print(f"Star {np.argmin(magnitude)}: magnitude = {min(magnitude)}") #Brightest star
print(f"Star {np.argmax(magnitude)}: magnitude = {max(magnitude)}") #Dimmest star

print(f"Star {np.argmin(temperature)}: temperature = {min(temperature)}")
print(f"Star {np.argmax(temperature)}: temperature = {max(temperature)}")
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.plot(figsize=(8,6))
plt.scatter(log_temperature, log_magnitude, label="Stars")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.scatter(log_temperature[np.argmin(log_magnitude)], min(log_magnitude), label="Brightest Star", color="black", marker="*", s=300)
plt.scatter(log_temperature[np.argmax(log_magnitude)], max(log_magnitude), label="Dimmest Star", color="purple", marker="*", s=300)
plt.scatter(max(log_temperature), log_magnitude[np.argmax(log_temperature)], label="Hottest Star", color="green", marker="*", s=300)
plt.scatter(min(log_temperature), log_magnitude[np.argmin(log_temperature)], label="Coldest Star", color="orange", marker="*", s=300)
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.legend()
```
