```python
arr = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])

def is_prime(i):
    if i < 2:
        return False
    for n in range(2, int(i**0.5) + 1):
        if i % n == 0:
            return False
    else:
        return True
        
def rows_with_primes(array):
    prime_rows = []
    for row in array:
        if any(is_prime(i) for i in row):
            prime_rows.append(row)
    return np.array(prime_rows)

print(rows_with_primes(arr))
```
